## Exercise 1 — Present Discounted Value

The present discounted value (PDV) of a future payment is

$$
PDV = \frac{FV}{(1+r)^t},
$$
where $FV$ is the future value, $r$ is the annual interest rate, and $t$ is the number of years.

Suppose you will receive $10,000 in 30 years, and the annual interest rate is 2.5%.

1. Define `FV = 10000`, `r = 0.025`, and `t = 30`. Calculate the PDV and save the result in a variable called `PDV`.


````python
# Variables for the future value calculation
FV = 10000
r = 0.025
t = 30

# Store PDV in a variable
PDV = FV / ((1 + r) ** t)

print(f"Your present discounted value is: ${PDV:.2f}")

# Extra challenge: Write a function instead

def future_value(FV, r, t):
    return FV / ((1 + r) ** t)

print(future_value(FV, r, t))


````
