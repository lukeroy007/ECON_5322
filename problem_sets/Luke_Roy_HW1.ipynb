## Exercise 1 — Present Discounted Value

The present discounted value (PDV) of a future payment is

$$
PDV = \frac{FV}{(1+r)^t},
$$
where $FV$ is the future value, $r$ is the annual interest rate, and $t$ is the number of years.

Suppose you will receive $10,000 in 30 years, and the annual interest rate is 2.5%.

1. Define `FV = 10000`, `r = 0.025`, and `t = 30`. Calculate the PDV and save the result in a variable called `PDV`.


````python
# Variables for the future value calculation
FV = 10000
r = 0.025
t = 30

# Store PDV in a variable
PDV = FV / ((1 + r) ** t)

print(f"Your present discounted value is: ${PDV:.2f}")

# Extra challenge: Write a function instead

def future_value(FV, r, t):
    return FV / ((1 + r) ** t)

print(future_value(FV, r, t))


````

## Exercise 2 — A Log Approximation

For two positive numbers $x$ and $y$ that are close to each other,

$$
\log(x)-\log(y) \approx \frac{x-y}{y}.
$$

Use Python's `math` module (to get log value) for this exercise.

1. Let `x = 101` and `y = 100`. Compute both sides of the approximation and compare them.


````python
from math import log

# Define variables

x = 101 
y = 100

log_var = log(x) - log(y)
div_var = (x - y) / y

print(log_var)
print(div_var)


# They are very close to each other
````

2. Calculate the absolute difference between the two values.


````python
print(log_var - div_var) # Close to 0
````

3. Now consider `x = 5` and `y = 4`. **Before computing**, predict whether the approximation will be better or worse than in part 1. Briefly explain your reasoning.
