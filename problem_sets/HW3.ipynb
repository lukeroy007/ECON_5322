### Exercise 9

In this example, you will simulate and plot the following AR(1) process:

$$
x_{t+1} = \alpha x_{t} + \epsilon_{t+1},
$$

where $x_{0} = 0$ and $t = 0, ..., T$.

The sequence of shocks $\{\epsilon_{t}\}$ is assumed to be IID and standard normal (hint: use `np.random.randn()`).

Define $T = 200$ and consider the values:

$\alpha = 0.0$, $\alpha = 0.7$, and $\alpha = 0.98$

1. On the same figure, plot all three simulated time series.

Use a **for** loop to step through the $\alpha$ values. Hint: You will need a nested for loop. The inner loop generates each AR(1) time series, while the outer loop iterates over the different values of $\alpha$. 
You don't need to add a legend.

First, you need to import the following packages:


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
T = 200
alphas = [0.0, 0.7, 0.98]

# Main loop

for alpha in alphas:
    x = [0]
    
    for t in range(T):
        x_next = alpha * x[-1] + np.random.randn()
        x.append(x_next)
    plt.plot(x, label=f"alpha = {alpha}")



# Readability

plt.legend()
plt.xlabel("Time")
plt.ylabel("x")
plt.title("Simulated AR(1) Processes")
plt.legend()
plt.show()

```

2. How does the behavior of the time series change as alpha increases? Which value of alpha produces the most persistent series, and why?

Answer:

#### As alpha gets closer and closer to 1, the series becomes more persistent. This is because alpha controls how much of the previous period's values survive into the next period. So if $a$ = .98, then 98% of the previous value is carried into the new value, plus whatever new shock occurs. That is why $a$ = .98 is the most persistent.
