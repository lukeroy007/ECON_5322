### A small note about `zip()` and `enumerate()`

`zip()` and `enumerate()` return iterable objects. An important quirk of some iterable types that are not lists (such as  `zip`) is that <u>you cannot convert the same type to a list twice.</u>

This is because `zip` and `enumerate` produce what is called a generator.

A generator will only produce each of its elements a single time, so if you call `list` on the same
generator a second time, it will not have any elements to iterate over anymore.

# 5. Sets

Sets are less common in this course, but you should be aware of them.

A set is an **unordered collection of unique elements**. Sets are closely related to the mathematical concept of a set.

We can create sets using the following process with **braces**:

`{item1, item2, ..., itemN}`

Unlike lists and tuples, elements of a set cannot be extracted using `s[N]`. It is because <u>sets are not ordered</u>, hence the notion of getting an element of a set is not well defined.
