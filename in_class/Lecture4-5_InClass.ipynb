## 1. Conditional statements

In programming languages, the conditional statement is one of the most useful techniques.

Boolean expressions produce `True` or `False`. Conditional statements use those results to control which block of code is executed.


### Comparison operators

- `x != y` : x is not equal to y
- `x > y` : x is greater than y
- `x < y` : x is less than y
- `x >= y` : x is greater than or equal to y
- `x <= y` : x is less than or equal to y

**Remember:** `=` assigns a value, while `==` compares values.

`is` and `is not` test whether two names refer to the same object; they are **not** substitutes for `==` when comparing values.


### The basic `if` statement

If we want to execute code only when a condition is true, we use an `if` statement.

```python
if condition:
    # code to run when condition is True
else:
    # code to run when no condition above are True
```

Notice the **colon** and the **indentation**.


The `else` clause is optional. If the condition is false, the indented block under `if` is skipped.

## 2. Iteration

In economics and data analysis, we often repeat a calculation for many observations or periods. This is called **iteration**.

Examples include computing revenue and cost over time, simulating a time series, or searching for the first value that satisfies a condition.


### `for` loops

A `for` loop repeats a block of code for each item in an iterable.


Suppose we want the squares of the first few integers. A loop lets us write the operation once and repeat it.


Useful when we have iterable `i` from 1 to 100!

### General structure


```python
for item in iterable:
    # operation 1
    # operation 2
```

The `iterable` can be a list, tuple, dictionary, string, `range`, `zip`, or `enumerate` object. The loop variable takes one value at a time.

### `while` loops

A `while` loop repeats **as long as a condition remains true**.


```python
while True_condition:
    # repeat these steps
```

A common source of errors is forgetting to update the variable used in the condition. That can create an **infinite loop**.

For example, the following code never changes `n`, so the condition `n > 0` never becomes false:

```python
n = 5
while n > 0:
    print("Lather")
    print("Rinse")
```
